## 1. 检查环境

确认 GPU 和依赖库都就绪。

```python
import torch
import ultralytics
import cv2

print("torch:", torch.__version__)
print("CUDA 可用:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
print("ultralytics:", ultralytics.__version__)
print("opencv:", cv2.__version__)
```

## 2. 加载模型

首次运行会自动从网上下载预训练权重到 `D:\tobacco\weights`。

```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")
```

## 3. 查看模型结构

打印网络层，理解 YOLO 的 backbone / neck / head 三部分。

```python
print(model.model)
```

## 4. 单张图片推理可视化

把 `img_path` 改成你自己的图片路径，运行后会在下方显示带检测框的结果。

```python
import matplotlib.pyplot as plt
from PIL import Image

img_path = "D:/tobacco/datasets/你的图片.jpg"
img = Image.open(img_path)
results = model(img)
r = results[0]

plt.figure(figsize=(10, 10))
plt.imshow(r.plot()[:, :, ::-1])
plt.axis("off")
plt.show()
```

## 5. 查看检测到的目标信息

每个检测框的类别、置信度和坐标。

```python
for box in r.boxes:
    cls_id = int(box.cls[0])
    conf = float(box.conf[0])
    xyxy = box.xyxy[0].tolist()
    print(f"类别={model.names[cls_id]}  置信度={conf:.3f}  坐标={xyxy}")
```
